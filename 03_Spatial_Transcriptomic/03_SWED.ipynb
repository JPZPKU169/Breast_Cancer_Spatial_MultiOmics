# SWED (Spatially weighted expression domain)

```python
import scanpy as sc
from easySCFpy import loadH5
import numpy as np
import pandas as pd
#import squidpy as sq
from collections import Counter
import matplotlib.pyplot as plt
from easySCFpy import saveH5
```

## Plot SWED spatial distributions in a loop

```python
sample_list_sm_matched = [
    "P1","P2","P3","P4","P5","P6","P7","P8","P9","P10","P11","P12","P14","P15","P16","P26","P27","P30","P32","P33","P34","P35","P36","P37","P38","P39",
    "P40_1","P40_2","P42","P43_1","P43_2","P44","P45","P46","P47","P48","P49"
]
```

```python
for spID in sample_list_sm_matched:
    st_obj = loadH5(f"E:/Projects/BRCA/Intermediate_Data/ST/WNN_objects/{spID}_epi_wnn.h5", image_name = "slice1",  assay = "Spatial")
    
    sq.gr.spatial_neighbors(st_obj, n_neighs=20, coord_type='generic')
    
    expr_adjacency = st_obj.obsp['wsnn']
    spatial_adjacency = st_obj.obsp['spatial_connectivities']
    
    alpha = 0.7
    beta = 0.3
    integrated_adjacency = alpha * expr_adjacency + beta * spatial_adjacency
    
    sc.tl.louvain(st_obj, key_added='sweds', resolution=0.3, adjacency=integrated_adjacency)
    
    fig = plt.figure(figsize=(7.87, 7.87), dpi=600)
    sc.pl.spatial(st_obj, color='sweds', spot_size=20, show=False, alpha_img=0)
    plt.savefig(f"E:/Projects/BRCA/results/ST/WNN/{spID}/SWED_alpha0.7_beta0.3_res0.3.png", dpi=600, bbox_inches='tight')
    
    sc.tl.louvain(st_obj, key_added='sweds', resolution=0.5, adjacency=integrated_adjacency)
    fig = plt.figure(figsize=(7.87, 7.87), dpi=600)
    sc.pl.spatial(st_obj, color='sweds', spot_size=20, show=False, alpha_img=0)
    plt.savefig(f"E:/Projects/BRCA/results/ST/WNN/{spID}/SWED_alpha0.7_beta0.3_res0.5.png", dpi=600, bbox_inches='tight')
    
    sc.tl.louvain(st_obj, key_added='sweds', resolution=1, adjacency=integrated_adjacency)
    fig = plt.figure(figsize=(7.87, 7.87), dpi=600)
    sc.pl.spatial(st_obj, color='sweds', spot_size=20, show=False, alpha_img=0)
    plt.savefig(f"E:/Projects/BRCA/results/ST/WNN/{spID}/SWED_alpha0.7_beta0.3_res1.png", dpi=600, bbox_inches='tight')
    
```
